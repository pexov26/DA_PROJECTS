## 1. Study vs Sleep Analysis


```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x='study_hours_per_day', y='sleep_hours', data=df)
plt.show()
```

### Insight 1: Study vs Sleep Relationship

From the scatter plot, there appears to be no clear relationship between study hours and sleep hours.

The data points are randomly distributed, indicating that students who study more do not necessarily sleep less or more.

This suggests that sleep patterns are independent of study time in this dataset.



## 2. Correlation Analysis


```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

### Insight 2: Factors Affecting Productivity

The correlation heatmap reveals several key relationships with productivity:

- Study hours per day shows a strong positive correlation (0.73) with productivity, indicating that increased study time significantly improves productivity.
- Focus score also has a moderate positive correlation (0.41), suggesting that concentration plays an important role.
- Sleep hours (0.34) contribute positively, showing that proper rest supports productivity.
- Phone usage hours have a negative correlation (-0.33), indicating that excessive phone use reduces productivity.
- Stress level shows a slight negative impact (-0.20), suggesting higher stress lowers productivity.

Overall, productivity is influenced by a combination of study habits, focus, sleep, and distractions.


## 3. Distraction Analysis


```python
df['distraction_score'] = (
    df['phone_usage_hours'] +
    df['social_media_hours'] +
    df['youtube_hours'] +
    df['gaming_hours']
)

df[['distraction_score']].head()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='distraction_score',
    y='productivity_score',
    data=df,
    alpha=0.3   # makes points transparent → easier to see pattern
)

plt.title("Distraction vs Productivity")
plt.xlabel("Distraction Score")
plt.ylabel("Productivity Score")

plt.show()

```

### Insight 3: Impact of Distraction on Productivity

The scatter plot shows a weak negative relationship between distraction score and productivity.

As distraction increases, productivity tends to decrease, although the pattern is not very strong and there is significant variation.

This suggests that while distractions like phone usage, social media, and gaming can reduce productivity, other factors also influence student performance.


## 4. Performance Segmentation


```python
df['performance_group'] = df['productivity_score'].apply(
    lambda x: 'High' if x >= df['productivity_score'].median() else 'Low'
)

df[['productivity_score', 'performance_group']].head()
```

```python
df.groupby('performance_group').mean(numeric_only=True)
```

### Insight 4: High vs Low Performing Students

By comparing high and low productivity students, clear behavioral differences emerge:

- High-performing students tend to study more hours per day.
- They have lower distraction scores, indicating reduced time spent on phone usage, social media, and gaming.
- They maintain better focus levels compared to low-performing students.
- Low-performing students show higher distraction and slightly lower study time.

This analysis highlights that productivity is strongly influenced by study habits, focus, and distraction levels.
