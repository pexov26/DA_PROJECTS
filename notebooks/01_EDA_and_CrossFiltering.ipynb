```python
import pandas as pd
import plotly.express as px

# Load the data (the '../' tells it to go up one folder to find the 'data' directory)
df = pd.read_csv('../data/ultimate_student_metrics_30k.csv')

# Let's check the first 5 rows to confirm it loaded safely
df.head()
```

```python
fig = px.histogram(data_frame=df, x="Exam_Score", title="Student Grade Distribution")
fig.show()
```

```python
fig_scatter = px.scatter(data_frame=df, x="Phone_Usage_Hours_Daily", y="Exam_Score", title="Impact of Phone Usage on Exam Scores")
fig_scatter.show()
```
